```python
# 1. Завантаження набору даних Iris
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis

# Завантажуємо дані
iris = load_iris()
X = iris.data
y = iris.target
print("Форма даних:", X.shape)
print("Класи:", np.unique(y))

# 2. Розподіл на train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
print("Train:", X_train.shape, "Test:", X_test.shape)

# 3. Вибірка ознак окремо для кожного класу
classes = np.unique(y_train)
class_data = {c: X_train[y_train == c] for c in classes}

# 4. Матриці коваріації
cov_matrices = {c: np.cov(class_data[c].T) for c in classes}

# 5. Обернені матриці коваріації
inv_cov_matrices = {c: np.linalg.inv(cov_matrices[c]) for c in classes}

# 6. Апріорні ймовірності
priors = {c: len(class_data[c]) / len(X_train) for c in classes}
print("Апріорні ймовірності:", priors)

# 7. Функція дискримінантної функції для одного вектора
def discriminant_function(x, mean, inv_cov, prior):
    diff = x - mean
    return -0.5 * diff @ inv_cov @ diff.T + np.log(prior)

# 8. Функція для всієї матриці тестових даних
def predict_custom(X_test):
    predictions = []
    for x in X_test:
        scores = {}
        for c in classes:
            mean = np.mean(class_data[c], axis=0)
            scores[c] = discriminant_function(x, mean, inv_cov_matrices[c], priors[c])
        predictions.append(max(scores, key=scores.get))
    return np.array(predictions)

y_pred_custom = predict_custom(X_test)

# 9. Прогнозування через QuadraticDiscriminantAnalysis
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train, y_train)
y_pred_qda = qda.predict(X_test)

# 10. Порівняння результатів
accuracy_custom = np.mean(y_pred_custom == y_test)
accuracy_qda = np.mean(y_pred_qda == y_test)

print("Точність власної реалізації:", accuracy_custom)
print("Точність sklearn QDA:", accuracy_qda)

```
